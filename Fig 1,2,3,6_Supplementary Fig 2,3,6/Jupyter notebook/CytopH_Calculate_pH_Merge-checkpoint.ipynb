Calculate baseline pH assuming pH = 7.4 for peak intensity on NH4Cl addition

```python
from pathlib import Path
import numpy as np
import pandas as pd

# ============================================================
# INPUT
# ============================================================
input_dir = Path("../input_images_subfolder 2D BgS")

# average first x frames for each Spine_ID to define baseline
baseline_n_frames = 3

# pH of NH4Cl condition
peak_pH_value = 7.4

# pKa of the probe
# change this if needed for your construct / calibration
pKa = 7.1

# file pattern
file_pattern = "Dish*_F.csv"

# output suffix
output_suffix = "_pH.csv"


# ============================================================
# FUNCTIONS
# ============================================================
def compute_baseline_per_spine(df: pd.DataFrame, baseline_n_frames: int) -> pd.DataFrame:
    """
    For each Spine_ID, compute baseline intensity as the average
    of the first baseline_n_frames frames.
    """
    df = df.sort_values(["Spine_ID", "Frame"]).copy()

    baseline_df = (
        df.groupby("Spine_ID", group_keys=False)
          .head(baseline_n_frames)
          .groupby("Spine_ID", as_index=False)["Intensity"]
          .mean()
          .rename(columns={"Intensity": "baseline_int"})
    )

    return baseline_df


def compute_average_trace(df: pd.DataFrame) -> pd.DataFrame:
    """
    Average Intensity across Spine_ID for each Frame to create
    a single average trace for the file.
    """
    avg_trace = (
        df.groupby("Frame", as_index=False)["Intensity"]
          .mean()
          .rename(columns={"Intensity": "avg_intensity"})
          .sort_values("Frame")
          .reset_index(drop=True)
    )
    return avg_trace


def compute_peak_from_average_trace(avg_trace: pd.DataFrame, baseline_n_frames: int) -> float:
    """
    Find the peak in the average trace starting after the baseline window.
    If baseline_n_frames = 3, search frames 4-n.
    If baseline_n_frames = 9, search frames 10-n.

    This assumes Frame numbering is 1-based.
    """
    peak_search_start = baseline_n_frames + 1

    peak_region = avg_trace[avg_trace["Frame"] >= peak_search_start].copy()
    if peak_region.empty:
        raise ValueError(
            f"No frames available for peak search. "
            f"baseline_n_frames={baseline_n_frames}, "
            f"max frame in file={avg_trace['Frame'].max()}"
        )

    peak_int = peak_region["avg_intensity"].max()
    return float(peak_int)


def baseline_ph_from_peak_and_baseline(
    baseline_int: pd.Series,
    peak_int: float,
    pKa: float,
    peak_pH_value: float
) -> pd.Series:
    """
    Compute baseline pH from baseline fluorescence and fluorescence at known pH.

    Uses:
    F_pH7.4 = F_base * (1 + 10^(pKa - pH_base)) / (1 + 10^(pKa - 7.4))

    Solving for pH_base:
    pH_base = pKa - log10( (  (1 + 10^(pKa - 7.4)) / (F_base/F_pH7.4 )  )  - 1 )
    """
    baseline_int = baseline_int.astype(float)

    term = ( (1 + 10**(pKa - peak_pH_value))/(baseline_int/peak_int) ) - 1

    baseline_pH = np.where(
        (baseline_int > 0) & (term > 0),
        pKa - np.log10(term),
        np.nan
    )

    return pd.Series(baseline_pH, index=baseline_int.index)


# ============================================================
# MAIN
# ============================================================
csv_files = sorted(input_dir.glob(file_pattern))

if len(csv_files) == 0:
    raise FileNotFoundError(f"No files found matching {file_pattern} in {input_dir}")

for csv_path in csv_files:
    print(f"\nProcessing: {csv_path.name}")

    df = pd.read_csv(csv_path)

    required_cols = {"Spine_ID", "Frame", "Intensity"}
    missing = required_cols - set(df.columns)
    if missing:
        raise ValueError(f"{csv_path.name} is missing columns: {missing}")

    df = df[["Spine_ID", "Frame", "Intensity"]].copy()

    df["Frame"] = pd.to_numeric(df["Frame"], errors="coerce")
    df["Intensity"] = pd.to_numeric(df["Intensity"], errors="coerce")
    df = df.dropna(subset=["Spine_ID", "Frame", "Intensity"]).copy()

    baseline_df = compute_baseline_per_spine(df, baseline_n_frames)
    avg_trace = compute_average_trace(df)
    peak_int = compute_peak_from_average_trace(avg_trace, baseline_n_frames)

    out_df = baseline_df.copy()
    out_df["peak_int"] = peak_int
    out_df["baseline_pH"] = baseline_ph_from_peak_and_baseline(
        baseline_int=out_df["baseline_int"],
        peak_int=peak_int,
        pKa=pKa,
        peak_pH_value=peak_pH_value
    )
    out_df["peak_pH"] = peak_pH_value

    out_df = out_df[["Spine_ID", "baseline_int", "peak_int", "baseline_pH", "peak_pH"]]

    # round to 3 decimal places
    out_df[["baseline_int", "peak_int", "baseline_pH", "peak_pH"]] = (
        out_df[["baseline_int", "peak_int", "baseline_pH", "peak_pH"]].round(3)
    )

    out_path = csv_path.with_name(csv_path.stem + output_suffix)
    out_df.to_csv(out_path, index=False)

    print(f"Saved: {out_path.name}")
    print(f"Peak intensity used for all spines in this file: {peak_int:.3f}")

print("\nDone.")
```

Merge CSVs

```python
from pathlib import Path
import re
import pandas as pd

# ============================================================
# INPUT
# ============================================================
input_dir = Path("../input_images_subfolder 2D BgS")
output_dir = Path("../Measurements")
output_dir.mkdir(parents=True, exist_ok=True)

file_pattern = "Dish*_F_pH.csv"

# names to detect in the input folder name only
condition_names = ["WT", "VAPBP56S", "control shRNA", "MPC1", "PGK1"]


# ============================================================
# DETERMINE OUTPUT NAME FROM FOLDER NAME ONLY
# ============================================================
folder_name = input_dir.name

matched_condition = None
for name in condition_names:
    if name.lower() in folder_name.lower():
        matched_condition = name
        break

if matched_condition is None:
    raise ValueError(
        f"None of {condition_names} were found in folder name: {folder_name}"
    )

output_csv = output_dir / f"{matched_condition}_pH.csv"


# ============================================================
# FIND FILES
# ============================================================
csv_files = sorted(input_dir.glob(file_pattern))

if len(csv_files) == 0:
    raise FileNotFoundError(f"No files found matching {file_pattern} in {input_dir}")


# ============================================================
# HELPER TO EXTRACT DISH NUMBER
# ============================================================
def extract_dish_number(filename: str):
    """
    Extract number from names like:
    Dish1-1_F_pH.csv -> 1
    Dish2-1_F_pH.csv -> 2
    Dish12_F_pH.csv  -> 12
    """
    match = re.search(r"Dish(\d+)", filename, flags=re.IGNORECASE)
    if match:
        return int(match.group(1))
    return pd.NA


# ============================================================
# MERGE
# ============================================================
all_dfs = []

for csv_path in csv_files:
    print(f"Reading: {csv_path.name}")
    df = pd.read_csv(csv_path)

    required_cols = {"Spine_ID", "baseline_pH"}
    missing = required_cols - set(df.columns)
    if missing:
        raise ValueError(f"{csv_path.name} is missing columns: {missing}")

    dish_number = extract_dish_number(csv_path.name)

    out_df = df[["Spine_ID", "baseline_pH"]].copy()
    out_df.insert(0, "Dish", dish_number)

    all_dfs.append(out_df)

merged_df = pd.concat(all_dfs, ignore_index=True)

# keep only requested columns in requested order
merged_df = merged_df[["Dish", "Spine_ID", "baseline_pH"]]

# optional sort
merged_df = merged_df.sort_values(["Dish", "Spine_ID"]).reset_index(drop=True)

# save
merged_df.to_csv(output_csv, index=False)

print(f"\nSaved merged CSV: {output_csv}")
print(f"Rows: {len(merged_df)}")
```
